```python
import pandas as pd

day1 = pd.read_parquet("data/training/checkpoints/combined_days/20260620_combined.parquet")  # day BEFORE the flare
day2 = pd.read_parquet("data/training/checkpoints/combined_days/20260621_combined.parquet")  # the flare day itself

combined = pd.concat([day1, day2])
combined.to_parquet("june21_with_leadin.parquet")
```

```python
combined = pd.read_parquet("june21_with_leadin.parquet")
window = combined.loc["2026-06-21 19:45:00":"2026-06-21 20:15:00"]
print("hel1os_ctr == 0 fraction:", (window["hel1os_ctr"] == 0).mean())
print("hel1os_err == 0 fraction:", (window["hel1os_err"] == 0).mean())
```
